```python
median_before = df['horsepower'].median()
most_frequent = df['horsepower'].mode()[0]
df['horsepower'] = df['horsepower'].fillna(most_frequent)
median_after = df['horsepower'].median()
print(median_before, median_after)
```

```python
asia_df = df[df['origin'] == 'Asia']
X = asia_df[['vehicle_weight', 'model_year']].iloc[:7].values
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])  # check the homework page for the exact y values given
w = XTX_inv.dot(X.T).dot(y)
print(w.sum())
```
